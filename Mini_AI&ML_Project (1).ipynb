```python
class StudentReport:
  def __init__(self, name, age, skills):
    self.name = name
    self.age = age
    self.skills = skills

  def strongest_skill(self):
    if not self.skills:
      return "No skills available"
    return max(self.skills, key=self.skills.get)

  def weakest_skill(self):
    if not self.skills:
      return "No skills available"
    return min(self.skills, key=self.skills.get)

  def learning_recommendation(self, skill, score):
    if score < 60:
      return f"Improve your fundamentals in {skill}."
    elif score < 80:
      return f"Practice intermediate projects in {skill}."
    else:
      return f"Work on advanced projects in {skill}."

  def calculate_average(self):
    if not self.skills:
      return 0.0
    return sum(self.skills.values()) / len(self.skills)

  def determine_level(self):
    average = self.calculate_average()
    if average >= 80:
      return "Advanced"
    elif average >= 60:
      return "Intermediate"
    else:
      return "Beginner"

  def generate_report(self):
    average = self.calculate_average()
    level = self.determine_level()
    strongest = self.strongest_skill()
    weakest = self.weakest_skill()
    recommendation = self.learning_recommendation(weakest, self.skills[weakest])

    print("\n" + "=" * 50)
    print("AI STUDENT PERFORMANCE REPORT")
    print("=" * 50)

    print(f"Name: {self.name}")
    print(f"Age: {self.age}")

    print("\nSkills:")
    for skill, score in self.skills.items():
      print(f"  {skill}: {score}")

    print(f"\nAverage Score: {average:.2f}")
    print(f"Skill Level: {level}")
    print(f"Strongest Skill: {strongest}")
    print(f"Skill to Improve: {weakest}")

    print("\nRecommendation:")
    print(f"  {recommendation}")

    print("=" * 50)

# Example usage:
student_skills = {"python": 85, "sql": 72, "statistics": 60, "machine learning": 55}
student = StudentReport("John Doe", 22, student_skills)
student.generate_report()
```

```python
class AIStudent(StudentReport):

    def __init__(self, name, age, skills, specialization):
        super().__init__(name, age, skills)

        self.specialization = specialization

    # Override generate_report to include specialization
    def generate_report(self):

        average = self.calculate_average()
        level = self.determine_level()
        strongest = self.strongest_skill()
        weakest = self.weakest_skill()

        recommendation = self.learning_recommendation(
            weakest,
            self.skills[weakest]
        )

        print("\n" + "=" * 55)
        print("          AI STUDENT PERFORMANCE REPORT")
        print("=" * 55)

        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Specialization: {self.specialization}")

        print("\nSkills:")

        for skill, score in self.skills.items():
            print(f"  {skill}: {score}")

        print(f"\nAverage Score: {average:.2f}")
        print(f"Skill Level: {level}")
        print(f"Strongest Skill: {strongest}")
        print(f"Skill to Improve: {weakest}")

        print("\nRecommendation:")
        print(f"  {recommendation}")

        print("=" * 55)
student1 = AIStudent(
    "Abdulsalam",
    20,
    {
        "Python": 85,
        "SQL": 72,
        "Statistics": 60,
        "Machine Learning": 55
    },
    "Machine Learning"
)


student2 = AIStudent(
    "Sarah",
    23,
    {
        "Python": 90,
        "SQL": 85,
        "Statistics": 88,
        "Machine Learning": 82
    },
    "Data Science"
)


student3 = AIStudent(
    "Michael",
    21,
    {
        "Python": 55,
        "SQL": 48,
        "Statistics": 52,
        "Machine Learning": 45
    },
    "AI Engineering"
)


# Store students inside a list
students = [
    student1,
    student2,
    student3
]
students

print("\n\n")
print("#" * 60)
print("                 ALL STUDENT REPORTS")
print("#" * 60)

for student in students:
    student.generate_report()
```
