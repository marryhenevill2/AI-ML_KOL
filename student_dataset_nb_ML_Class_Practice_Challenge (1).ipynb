## 8. Preprocessing

Prepare numerical and categorical variables.

Consider:
- Missing-value imputation
- Standardization/scaling of numerical features where appropriate
- One-hot encoding of categorical features

Use a **Pipeline** and **ColumnTransformer** where appropriate so that preprocessing is learned only from the training data.

### Question
Why is it dangerous to calculate preprocessing statistics using the complete dataset before the train/test split?

```python
# TODO: Identify numerical and categorical features

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_features = X.select_dtypes(include=np.number).columns.tolist()
categorical_features = X.select_dtypes(exclude=np.number).columns.tolist()

print("Numerical features:", numeric_features)
print("Categorical features:", categorical_features)

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```
