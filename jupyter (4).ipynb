```python
# TASK 1 - BASIC STUDENT INFORMATION

student_name = "Abdulsalam"
age = 20
is_enrolled = True
specialization = "Machine Learning"

# Demonstrating different Python data types
print("Student Name:", student_name)
print("Age:", age)
print("Enrolled:", is_enrolled)
print("Specialization:", specialization)

# TASK 2 - SKILLS DICTIONARY

skills = {
    "Python": 85,
    "SQL": 72,
    "Statistics": 60,
    "Machine Learning": 55
}

# TASK 3 - OTHER DATA STRUCTURES
# LIST
student_names = ["Abdulsalam", "Sarah", "Michael", "Grace"]

# TUPLE
course_info = ("AI/ML", "6 Months")

# SETS
student1_skills = {"Python", "SQL", "Machine Learning"}
student2_skills = {"Python", "Statistics", "Machine Learning"}

# Find common skills
common_skills = student1_skills.intersection(student2_skills)

print("\nCommon Skills:", common_skills)

# TASK 4 - CALCULATE AVERAGE
scores = [20, 30, 60, 50, 70, 90]
def calculate_average(scores):
    if not scores:
        return 0

    return sum(scores) / len(scores)
average = calculate_average(scores)
print(average)

# TASK 5 - DETERMINE SKILL LEVEL

def determine_level(score):
    """
    Determine student level based on average score.
    """

    if score >= 80:
        return "Advanced"

    elif score >= 60:
        return "Intermediate"

    else:
        return "Beginner"

# TASK 6 - FIND STRONGEST SKILL

def strongest_skill(skills):
    """
    Return the skill with the highest score.
    """

    if not skills:
        return "No skills available"

    return max(skills, key=skills.get)


# ============================================================
# TASK 7 - FIND WEAKEST SKILL
# ============================================================

def weakest_skill(skills):
    """
    Return the skill with the lowest score.
    """

    if not skills:
        return "No skills available"

    return min(skills, key=skills.get)


# ============================================================
# TASK 8 - LEARNING RECOMMENDATION
# ============================================================

def learning_recommendation(skill, score):
    """
    Generate a learning recommendation based on score.
    """

    if score < 60:
        return f"Improve your fundamentals in {skill}."

    elif score < 80:
        return f"Practice intermediate projects in {skill}."

    else:
        return f"Work on advanced projects in {skill}."


# ============================================================
# TASK 9 - STUDENT CLASS
# ============================================================

def learning_recommendation(skill, score):
    """
    Generate a learning recommendation based on score.
    """

    if score < 60:
        return f"Improve your fundamentals in {skill}."

    elif score < 80:
        return f"Practice intermediate projects in {skill}."

    else:
        return f"Work on advanced projects in {skill}."


    # ========================================================
    # TASK 10 - STUDENT METHODS
    # ========================================================

    def calculate_average(self):
        """Calculate the student's average score."""

        return calculate_average(self.skills.values())

    def determine_level(self):
        """Determine the student's overall level."""

        average = self.calculate_average()

        return determine_level(average)

    def strongest_skill(self):
        """Find the student's strongest skill."""

        return strongest_skill(self.skills)

    def weakest_skill(self):
        """Find the student's weakest skill."""

        return weakest_skill(self.skills)

    # ========================================================
    # TASK 12 - GENERATE STUDENT REPORT
    # ========================================================

    def generate_report(self):

        average = self.calculate_average()
        level = self.determine_level()
        strongest = self.strongest_skill()
        weakest = self.weakest_skill()

        recommendation = learning_recommendation(
            weakest,
            self.skills[weakest]
        )

        print("\n" + "=" * 50)
        print("AI STUDENT PERFORMANCE REPORT")
        print("=" * 50)

        print(f"Name: {self.name}")
        print(f"Age: {self.age}")

        print("\nSkills:")

        for skill, score in self.skills.items():
            print(f"  {skill}: {score}")

        print(f"\nAverage Score: {average:.2f}")
        print(f"Skill Level: {level}")
        print(f"Strongest Skill: {strongest}")
        print(f"Skill to Improve: {weakest}")

        print("\nRecommendation:")
        print(f"  {recommendation}")

        print("=" * 50)

# TASK 11 - AI STUDENT CLASS

class AIStudent(Student):

    def __init__(self, name, age, skills, specialization):
        super().__init__(name, age, skills)

        self.specialization = specialization

    # Override generate_report to include specialization
    def generate_report(self):

        average = self.calculate_average()
        level = self.determine_level()
        strongest = self.strongest_skill()
        weakest = self.weakest_skill()

        recommendation = learning_recommendation(
            weakest,
            self.skills[weakest]
        )

        print("\n" + "=" * 55)
        print("          AI STUDENT PERFORMANCE REPORT")
        print("=" * 55)

        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Specialization: {self.specialization}")

        print("\nSkills:")

        for skill, score in self.skills.items():
            print(f"  {skill}: {score}")

        print(f"\nAverage Score: {average:.2f}")
        print(f"Skill Level: {level}")
        print(f"Strongest Skill: {strongest}")
        print(f"Skill to Improve: {weakest}")

        print("\nRecommendation:")
        print(f"  {recommendation}")

        print("=" * 55)


# TASK 13 - CREATE MULTIPLE AI STUDENTS\

student1 = AIStudent(
    "Abdulsalam",
    20,
    {
        "Python": 85,
        "SQL": 72,
        "Statistics": 60,
        "Machine Learning": 55
    },
    "Machine Learning"
)


student2 = AIStudent(
    "Sarah",
    23,
    {
        "Python": 90,
        "SQL": 85,
        "Statistics": 88,
        "Machine Learning": 82
    },
    "Data Science"
)


student3 = AIStudent(
    "Michael",
    21,
    {
        "Python": 55,
        "SQL": 48,
        "Statistics": 52,
        "Machine Learning": 45
    },
    "AI Engineering"
)


# Store students inside a list
students = [
    student1,
    student2,
    student3
]
students

# TASK 14 - ANALYZE ALL STUDENTS

print("\n\n")
print("#" * 60)
print("             ALL STUDENT REPORTS")
print("#" * 60)

for student in students:
    student.generate_report()


# ============================================================
# TASK 15 - CLASSIFY STUDENTS
# ============================================================

def classify_students(students):
    """
    Organize students into Advanced, Intermediate
    and Beginner categories.
    """

    classification = {
        "Advanced": [],
        "Intermediate": [],
        "Beginner": []
    }

    for student in students:

        level = student.determine_level()

        classification[level].append(student.name)

    return classification


classification = classify_students(students)

print("\n")
print("=" * 50)
print("              STUDENT CLASSIFICATION")
print("=" * 50)

for level, names in classification.items():

    print(f"{level}: {names}")


# ============================================================
# BONUS 1 - SEARCH STUDENT
# ============================================================

def search_student(students, name):
    """
    Search for a student by name.
    """

    for student in students:

        if student.name.lower() == name.lower():
            return student

    return None


# Test student search
search_name = "Sarah"

found_student = search_student(students, search_name)

print("\n")
print("=" * 50)
print("              STUDENT SEARCH")
print("=" * 50)

if found_student:
    print(f"Student found: {found_student.name}")
    print(f"Specialization: {found_student.specialization}")
else:
    print("Student not found.")

# BONUS 2 - FIND TOP STUDENT

def top_student(students):
    """
    Find the student with the highest average score.
    """

    if not students:
        return None

    return max(students, key=lambda student: student.calculate_average())


best_student = top_student(students)

print("\n")
print("=" * 50)
print("                TOP STUDENT")
print("=" * 50)

if best_student:
    print(f"Name: {best_student.name}")
    print(f"Average: {best_student.calculate_average():.2f}")
    print(f"Level: {best_student.determine_level()}")
```
